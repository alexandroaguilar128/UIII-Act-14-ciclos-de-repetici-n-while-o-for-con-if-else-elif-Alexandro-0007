1. Crea una aplicación que pida un número y calcule su factorial (El factorial de un número es el producto de todos los enteros entre 1 y el propio número y se representa por el número seguido de un signo de exclamación. Por ejemplo 5! = 1x2x3x4x5=120).


```python
# Paso 1: Pedir el número al usuario y convertirlo a entero [3, 4]
numero = int(input("Ingresa un número para calcular su factorial: "))

# Paso 2: Guardar el valor original para la impresión final
valor_original = numero

# Paso 3: Inicializar la variable que guardará el resultado en 1 [1, 5]
factorial = 1

# Paso 4: Utilizar un ciclo para realizar las multiplicaciones sucesivas [5, 6]
# El ciclo se ejecutará mientras el número sea mayor que 0 [5, 7]
while numero > 0:
    factorial = factorial * numero # Multiplicación acumulativa [8]
    numero = numero - 1            # Decrementar el número en cada paso [5, 7]

# Paso 5: Mostrar el resultado final [4, 9]
print("El factorial de", valor_original, "es:", factorial)
```
